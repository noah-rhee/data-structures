### Asymptotic Analysis

**Asymptotic analysis** is a way of analyzing the performance of an algorithm by describing its behavior as the input size grows arbitrarily large. Instead of focusing on exact execution times, it provides a high-level understanding of the algorithm’s efficiency in terms of time or space complexity.

Key Concepts
-	1.	Input Size (n):
    -	The size of the input data is typically denoted as  n , and asymptotic analysis describes how the algorithm performs as  n  becomes very large.
-	2.	Growth Rates:
    -	Asymptotic analysis examines the dominant factors that affect an algorithm’s performance, ignoring constants and lower-order terms.
-	3.	Efficiency Metrics:
    -	Time Complexity: How the running time grows with the input size.
    -	Space Complexity: How memory usage grows with the input size.

````python
def two_sum(nums, target):
    
    seen = {} # hash map
    for i, num in enumerate(nums): # 1: O(n)
        diff = target - num # if your current number(=num) and any of previous number(=seen) add up to target, it means (target - num) must be in 'seen' hash. 
        
        if diff in seen: # 2: hash lookup: O(1). If you iterate all over the previous elements, O(n).
            return [seen[diff], i]
        
        seen[num] = i
    
    return None  

# O(n) x O(1) = O(n)

# Test
nums = [2, 7, 11, 15]
target = 9
print(two_sum(nums, target))  # Output: [0, 1]
````

### Q1. [242. Valid Anagram](https://leetcode.com/problems/valid-anagram/description/?envType=problem-list-v2&envId=hash-table)

````python
class Solution:
    def isAnagram(self, s: str, t: str) -> bool:
        s = sorted(s)
        t = sorted(t)
        if s == t:
            return True
        else:
            return False
````

### Q2. [349. Intersection of Two Arrays](https://leetcode.com/problems/intersection-of-two-arrays/description/?envType=problem-list-v2&envId=hash-table)

````python
class Solution:
    def intersection(self, nums1: List[int], nums2: List[int]) -> List[int]:
        l = []
        for i in range(len(nums1)):
            if nums1[i] in nums2:
                if nums1[i] not in l:
                    l.append(nums1[i])
        return l
````

### Q3. [409. Longest Palindrome](https://leetcode.com/problems/longest-palindrome/description/?envType=problem-list-v2&envId=hash-table)

````python
class Solution:
    def longestPalindrome(self, s: str) -> int:
        d = {}
        count = 0
        
        for i in s:
            if i in list(d.keys()):
                d[i] += 1
            else:
                d[i] = 1

        odd = 0
  

        for j in d.values():
            if j % 2 == 0:
                count += j
            else:
                if j > odd:
                    if odd != 0:
                        count -= 1
                    count += odd
                    odd = j
                    
                else:
                    count += j
                    count -= 1
                    
    
        count += odd
             

        return count
````
